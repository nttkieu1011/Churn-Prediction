### I. Final overview

#### 1. Model Selection

After experimenting with and comparing three algorithms — Logistic Regression, Random Forest, and HistGradientBoosting — HistGradientBoosting was selected as the primary model based on its Cross-validation performance.

| Model | CV PR-AUC (Mean) | CV ROC-AUC (Mean) |
|---|---:|---:|
| Logistic Regression | 0.0764 | 0.6978 |
| Random Forest | 0.0923 | 0.7176 |
| HistGradientBoosting | **0.0991** | 0.7088 |

HistGradientBoosting achieved the highest mean PR-AUC, making it a suitable choice for this churn prediction task, where the churn rate is low and the dataset is highly imbalanced. Although Random Forest achieved a higher ROC-AUC, HistGradientBoosting was prioritized because of its stronger performance in identifying churn cases, as reflected by PR-AUC.

#### 2. Final Test Performance

The HistGradientBoosting model was trained on the entire training set (June–July) and evaluated on an independent test set (August).

| Metric | Test Result |
|---|---:|
| PR-AUC | 0.0848 |
| ROC-AUC | 0.7134 |
| Test churn rate | 2.45% |

The results indicate that the model has some ability to distinguish and rank merchants according to their churn risk. While the test PR-AUC is lower than the Cross-validation average, the ROC-AUC remains relatively stable. This suggests that the model's generalization performance has some limitations, particularly in identifying merchants who are likely to churn.

#### 3. Business Impact – Top-K Evaluation

Instead of applying a fixed probability threshold, this project uses a Top-K ranking approach to prioritize merchants based on their churn scores. This approach enables the Customer Success (CS) team to focus its resources on merchants with the highest estimated churn risk, especially when customer outreach capacity is limited.

The evaluation results on the August 2026 test set are as follows:

| Top-K | Actual Churn | Precision@K | Recall@K |
|---|---:|---:|---:|
| Top 50 | 7 | 14% | 3.41% |
| Top 100 | 22 | 22% | 10.73% |
| Top 200 | 32 | 16% | 15.61% |

Key insights:

- Among the 100 merchants with the highest churn scores, 22 actually churned within the following 30 days.
- The churn rate within the Top 100 group was 22%, approximately nine times the overall churn rate of the test set (2.45%).
- Expanding the prioritized list to the Top 200 identified 32 churned merchants, accounting for 15.61% of all churn cases in the test set.

Overall, the model can help the CS team allocate its resources more effectively by prioritizing merchants with higher churn risk rather than treating all merchants equally.

**Note:** The churn score is used to rank merchants by risk and should not be interpreted directly as the actual probability of churn.

### II. Model Limitations

Although the model demonstrates potential in ranking merchants by churn risk, several limitations remain:

**1. Limited Feature Set**

The model currently uses 11 features, primarily reflecting merchants' operational scale, revenue, and search activity trends. However, several factors that may influence a merchant's decision to leave the platform have not been included, such as platform engagement, service usage history, customer support interactions, and operational issues.

**2. Limited Training Period**

The training data covers only June and July 2026, while the test set contains data from August 2026. This relatively short observation period may prevent the model from fully capturing seasonal patterns and changes in merchant behavior over time.

**3. Class Imbalance**

The churn rate in both the training and test sets is below 3%. The relatively small number of churn cases may limit the model's ability to learn patterns associated with churn and lead to variations in evaluation results across different data splits.

**4. Model Score Calibration**

The model uses `class_weight` to address class imbalance. Therefore, the current churn scores are more suitable for ranking merchants by risk than for estimating their actual churn probabilities. Probability calibration has not yet been performed or validated.

**5. Limited Business Validation**

The current evaluation is based on historical data. The model has not yet been deployed in an actual merchant retention workflow. As a result, its effectiveness in helping the Customer Success (CS) team retain at-risk merchants through proactive interventions has not yet been measured.

### III. Future Improvements

To improve predictive performance and practical applicability, the project could be further developed in the following directions:

- **Feature Engineering:** Incorporate additional features related to merchant engagement, activity frequency, customer support history, and service usage behavior.
- **More Historical Data:** Expand the training dataset to cover a longer period, allowing the model to capture long-term trends and seasonal patterns in merchant behavior.
- **Model Improvement:** Experiment with additional boosting algorithms such as XGBoost and LightGBM, while also evaluating churn score calibration techniques.
- **Business-oriented Evaluation:** Combine Precision@K and Recall@K with the actual customer outreach capacity of the CS team to determine an appropriate size for the prioritized merchant list.
- **Model Monitoring:** Monitor model performance on a monthly basis, detect changes in data distributions, and reassess or retrain the model when necessary.
- **Feedback Loop:** Collect feedback and intervention outcomes from the CS team to evaluate whether proactive engagement activities help reduce merchant churn.
